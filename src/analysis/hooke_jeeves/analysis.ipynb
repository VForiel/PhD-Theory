```python
import numpy as np
import phise
import metrics
import matplotlib.pyplot as plt
```

# Metrics definition :

Average depth
$$
\frac{\Sigma_i N_i}{B}
$$

Max depth
$$
\frac{\max_i N_i}{B}
$$

Average flux
$$
\Sigma_i N_i
$$

Max flux
$$
\max_i N_i
$$

```python
considered_metrics = {
    "sum(Ni)/B": metrics.average_depth,
    "max(Ni)/B": metrics.max_depth,
    "sum(Ni)": metrics.average_flux,
    "max(Ni)": metrics.max_flux,
    "sum(Ni^2/B^2)": metrics.distance_normalized,
    "sum(Ni^2)" : metrics.distance,
}

ctx = phise.examples.contexts.get_PHOB()
ctx.camera.ideal = True

data = metrics.generate_data(ctx, metrics=considered_metrics, samples=1000)
```

```python
fig, axes, df = metrics.plot_metric_evolution(data)
plt.show()
df
```

```python
fig, axes, df = metrics.plot_final_null_depth_distributions(data, bins=60)
plt.show()
df
```

```python
fig, axes, df = metrics.plot_final_metric_distributions(data, bins=60)
plt.show()
df
```
