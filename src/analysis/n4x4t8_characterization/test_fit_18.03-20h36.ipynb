```python
from src.N4x4_T8_characterisation import fit_model
from src.utils import latexio

data_path = "./highlighted_results/2026-03-18_b38b520/2_N4x4-T8_characterisation/systematic_scan.npz"
scan_data = fit_model.load_scan_data(data_path)
```

---
# MPE
$$
I = | M \cdot P \cdot E |^2
$$

```python
MPE_metadata = fit_model.fit_MPE(scan_data, return_metadata=True)
fit_model.plot_fit(scan_data, MPE_metadata['model'])
latexio.display_complex_matrix(MPE_metadata['M'], name="M")
latexio.display_complex_vector(MPE_metadata['Eon'], name="Eon")
latexio.display_complex_vector(MPE_metadata['Eoff'], name="Eoff")
```
