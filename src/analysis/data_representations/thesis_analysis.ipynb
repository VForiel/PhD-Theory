## Time serie

```python
import data_representations
import phise
import astropy.units as u
import numpy as np
import matplotlib.pyplot as plt
from phise import Companion

ctx = phise.examples.contexts.get_VLTI()
ctx.interferometer.λ = 1.65 * u.um
ctx.interferometer.chip.σ = np.zeros(14) * u.nm
ctx.target.companions = []
ctx.monochromatic = True
ctx.target.companions.append(
    Companion(
        c=1e-2,
        ρ=2 * u.mas,
        θ=3*np.pi/4 * u.rad,
    )
)

# ctx.target.companions.append(
#     Companion(
#         c=1e-8,
#         ρ=8 * u.mas,
#         θ=0 * u.rad,
#     )
# )
# ctx.calibrate_obs(1000)

data, data_ref = data_representations.time_evolution(ctx, n=60, map=np.median, save_as="generated/thesis/time_evolution.png")
```
